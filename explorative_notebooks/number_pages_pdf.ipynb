```python
from pathlib import Path
from pypdf import PdfReader
from collections import defaultdict

pdf_ordner = Path("/Users/florianscheiber/PycharmProjects/llm-climate-neutrality-review/inputs/fulltext_pdf/allPDF/included-after-stage-2")

seiten_statistik = defaultdict(list)

for pdf in pdf_ordner.glob("*.pdf"):
    try:
        reader = PdfReader(pdf)
        seitenanzahl = len(reader.pages)
        seiten_statistik[seitenanzahl].append(pdf.name)

    except Exception as e:
        print(f"Fehler bei {pdf.name}: {e}")

print("STATISTIK")
for seiten in sorted(seiten_statistik):
    print(f"{len(seiten_statistik[seiten])} PDFs mit {seiten} Seite(n)")

print("\nDETAILS")
for seiten in sorted(seiten_statistik):
    print(f"\n{seiten} Seite(n):")
    for datei in sorted(seiten_statistik[seiten]):
        print(f"  - {datei}")
```

```python
from pathlib import Path
from pypdf import PdfReader

folder = Path("/Users/florianscheiber/PycharmProjects/llm-climate-neutrality-review/inputs/fulltext_pdf/allPDF/included-after-stage-2")

main_pages = 0
supplement_pages = 0

for pdf in folder.glob("*.pdf"):
    try:
        n_pages = len(PdfReader(pdf).pages)

        name = pdf.name.lower()

        if "main" in name:
            main_pages += n_pages
        elif "supplement" in name:
            supplement_pages += n_pages

    except Exception as e:
        print(f"Fehler bei {pdf.name}: {e}")

print(f"Main PDFs: {main_pages} Seiten")
print(f"Supplement PDFs: {supplement_pages} Seiten")
print(f"Gesamt: {main_pages + supplement_pages} Seiten")
```

```python
from pathlib import Path
from collections import defaultdict
from pypdf import PdfReader
import re

folder = Path("/Users/florianscheiber/PycharmProjects/llm-climate-neutrality-review/inputs/fulltext_pdf/allPDF/included-after-stage-2")

pages_per_article = defaultdict(int)

for pdf in folder.glob("*.pdf"):
    match = re.match(r"^(\d{3})-", pdf.name)

    if not match:
        continue

    article_id = match.group(1)

    try:
        n_pages = len(PdfReader(pdf).pages)
        pages_per_article[article_id] += n_pages

    except Exception as e:
        print(f"Fehler bei {pdf.name}: {e}")

for article_id, pages in sorted(
    pages_per_article.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(f"{article_id}: {pages} Seiten")
```
